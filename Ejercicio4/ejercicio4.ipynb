```python
import re

palabras_reservadas = {
    "int": "entero",
    "float": "flotante",
    "double": "doble precisión",
    "char": "carácter",
    "if": "si",
    "else": "sino",
    "for": "para",
    "while": "mientras",
    "do": "hacer",
    "switch": "seleccionar",
    "case": "caso",
    "break": "romper",
    "continue": "continuar",
    "return": "retornar",
    "void": "vacío",
    "struct": "estructura",
    "typedef": "definir tipo",
    "enum": "enumeración",
    "const": "constante",
    "static": "estático",
    "extern": "externo",
    "sizeof": "tamaño de",
    "unsigned": "sin signo",
    "signed": "con signo",
    "long": "largo",
    "short": "corto",
    "goto": "ir a",
    "default": "por defecto",
    "union": "unión",
    "volatile": "volátil",
    "register": "registro",
    "auto": "automático",
}


def traducir_palabras_reservadas(codigo):
    traducciones = {}
    for palabra in palabras_reservadas:
        patron = r"\b" + re.escape(palabra) + r"\b"
        if re.search(patron, codigo):
            traducciones[palabra] = palabras_reservadas[palabra]
    return traducciones


def main():
    ruta = "C:\\Users\\mleonet\\Desktop\\Lenguajes y Compiladores\\ejercicio2.c"
    try:
        with open(ruta, "r", encoding="utf-8") as archivo:
            codigo = archivo.read()
    except FileNotFoundError:
        print(f"No se encontró el archivo: {ruta}")
        return

    resultado = traducir_palabras_reservadas(codigo)

    if resultado:
        print("Palabras reservadas encontradas y su traducción:")
        for palabra, traduccion in resultado.items():
            print(f"{palabra} → {traduccion}")
    else:
        print("No se encontraron palabras reservadas en el código.")


if __name__ == "__main__":
    main()
```
