```python
# funcion para detectar notación FEN
# EJEMPLO DE UNA CADENA FEN -> rnbqkbnr/pppppppp/8/8/8/8/PPPPPPPP/RNBQKBNR w KQkq - 0 1


def isFenRow(s: str):
    admited = "rnbqkpRNBQKP12345678"
    total = 0

    for char in s:
        if char not in admited:
            return False
        if char.isdigit():
            total += int(char)
        else:
            total += 1

    if total != 8:
        return False
    return True


def isCastlingRight(s: str):
    admited = "KQkq"
    for char in s:
        if char not in admited:
            return False
    return True


def isFen(s: str):
    # Paso 1: verificar que haya exactamente 5 espacios
    if s.count(" ") != 5:
        return False

    # Paso 2: separar en campos
    subList = s.split(" ")
    print(subList)
    print("-" * 100)

    # Paso 3: recorrer cada campo por índice
    for i in range(len(subList)):
        campo = subList[i]

        if i == 0:
            # Validar posición de piezas
            if campo.count("/") != 7:
                return False
            filas = campo.split("/")
            for fila in filas:
                if not isFenRow(fila):
                    return False

        elif i == 1:
            # Validar turno
            if campo not in ["w", "b"]:
                return False

        elif i == 2:
            # Validar derechos de enroque
            if campo != "-" and not isCastlingRight(campo):
                return False

        elif i == 3:
            # Validar captura al paso
            if campo != "-":
                if len(campo) != 2:
                    return False
                if campo[0] not in "abcdefgh":
                    return False
                if campo[1] not in "36":
                    return False

        elif i == 4:
            # Validar medio movimiento
            if not campo.isdigit():
                return False
            if int(campo) < 0:
                return False

        elif i == 5:
            # Validar número de jugada
            if not campo.isdigit():
                return False
            if int(campo) < 1:
                return False
    return True
```

```python
message = "La cadena no es notacion FEN"
frase_fen = "rnbqkbnr/pppppppp/8/8/8/8/PPPPPPPP/RNBQKBNR w KQkq - 0 1"
isFen(frase_fen)
```
