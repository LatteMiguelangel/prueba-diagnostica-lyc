```python
def generar_triangulo_pascal(n):
    triangulo = []
    for i in range(n + 1):
        fila = [1] * (i + 1)
        for j in range(1, i):
            fila[j] = triangulo[i - 1][j - 1] + triangulo[i - 1][j]
        triangulo.append(fila)
    return triangulo[n]


def mostrar_polinomio(coeficientes):
    n = len(coeficientes) - 1
    partes = []
    for i, coef in enumerate(coeficientes):
        exp = n - i
        if exp == 0:
            partes.append(f"{coef}")
        elif exp == 1:
            partes.append(f"{coef}x")
        else:
            partes.append(f"{coef}x^{exp}")
    return " + ".join(partes)


def calcular_f_x(coeficientes, x):
    resultado = 0
    n = len(coeficientes) - 1
    for i, coef in enumerate(coeficientes):
        exp = n - i
        potencia = x**exp
        termino = coef * potencia
        print(f"Paso {i+1}: {coef} * {x}^{exp} = {termino}")
        resultado += termino
    print(f"Resultado final: f({x}) = {resultado}")
```

```python
n = 45
x = 10
coef = generar_triangulo_pascal(n)
print(f"(x+1)^{n} =", mostrar_polinomio(coef))
calcular_f_x(coef, x)
```
