```python
import re


def es_notacion_cientifica(cadena):
    patron = r"^[+-]?\d+(\.\d+)?[eE][+-]?\d+$"
    return re.fullmatch(patron, cadena) is not None


def es_direccion_ip(cadena):
    patron = (
        r"^((25[0-5]|2[0-4]\d|1\d\d|[1-9]?\d)\.){3}(25[0-5]|2[0-4]\d|1\d\d|[1-9]?\d)$"
    )
    return re.fullmatch(patron, cadena) is not None


def es_correo_electronico(cadena):
    patron = r"^[\w\.-]+@[\w\.-]+\.\w{2,}$"
    return re.fullmatch(patron, cadena) is not None


def reconocer_cadena(cadena):
    if es_notacion_cientifica(cadena):
        print("→ Es una expresión en notación científica.")
    elif es_direccion_ip(cadena):
        print("→ Es una dirección IP válida.")
    elif es_correo_electronico(cadena):
        print("→ Es un correo electrónico válido.")
    else:
        print("→ No coincide con ningún patrón reconocido.")


# Ejemplos de prueba
if __name__ == "__main__":
    ejemplos = [
        "1.23e+10",
        "192.168.0.1",
        "correo@ejemplo.com",
        "123abc",
        "3.14E-2",
        "256.100.50.25",
        "usuario@dominio",
    ]

    for cadena in ejemplos:
        print(f"Cadena: {cadena}")
        reconocer_cadena(cadena)
        print("-" * 40)
```
